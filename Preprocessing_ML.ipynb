```python
import pandas as pd

file_path = "/content/drive/MyDrive/TON_IoT_Project/train_test_network.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully")
print(df.shape)
```

```python
# Make a copy of original dataset
df_clean = df.copy()

print("Before duplicate removal:", df_clean.shape)

# Count duplicates
duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows found:", duplicate_count)

# Remove duplicates
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("After duplicate removal:", df_clean.shape)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# -----------------------------------
# 1. Target
# -----------------------------------
y = df_clean["label"]

# -----------------------------------
# 2. Use only already-numerical features
# -----------------------------------
X = df_clean.select_dtypes(include=np.number).copy()

# Remove target column from input
X = X.drop(columns=["label"], errors="ignore")

print("Number of features used:", X.shape[1])
print("Dataset shape for model:", X.shape)

# -----------------------------------
# 3. Train/Test split
# -----------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])

# -----------------------------------
# 4. Random Forest Model
# -----------------------------------
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

# -----------------------------------
# 5. Prediction
# -----------------------------------
y_pred = model.predict(X_test)

# -----------------------------------
# 6. Accuracy
# -----------------------------------
accuracy = accuracy_score(y_test, y_pred)

print("\nRandom Forest Accuracy:",
      round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
